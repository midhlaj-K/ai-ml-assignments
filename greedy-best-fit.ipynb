```python
from math import inf 

cost = {'AB': 4, 'AC': 3, 'BD': 2, 'BE': 1, 'CD': 1, 'CF': 2, 'DF': 1};
h = {'A': 10, 'B':6, 'C':8, 'D':7, 'E':11, 'F':0, '-':inf}; 
#eg: A
#heurinistic value - h['A']
#cost to B         - cost['AB'] 

neighbors = {'A': ['B', 'C'], 'B': ['A', 'D', 'E'], 'C': ['A','D', 'F'], 'D': ['B', 'C', 'F'], 'E': ['B'], 'F': ['D', 'C']}
```

Greedy Best Fit

```python
x = 0

def greedy(start, goal):


   # print(start,"  started")
    path = []
    path.append(start)

   # print(f"{start}------")
    ''''
    find_least() -> least
    if least == goal -> return path
    else path += (greedy(least) 
    '''

    #find least h(n)
    least = '-'
    for v in neighbors[start]:
        if h[v] < h[least]:
            least = v

   # print(f"least={least}, path-{path} \n")
    if (neighbors[start] == []) or (least=='-'):
        return []
    elif least == goal:
        return (path + [goal])

    else:
        path+=( greedy(least, goal))

    return path
        
    
gbf =(greedy('A', 'F'))
gbf
```
