Breadth First Search

```python

class edge:
  def __init__(self, start, end):
    self.source= start
    self.target = end

edgeList = (edge('A','B'), edge('B','D'), edge('D','G'), edge('B','E'), edge('A','C'), edge('C','F'), edge('F','G'))

def neighbors(v):
    nodes = []
    for e in edgeList:
        if e.source == v:
            nodes.append(e.target)
    if nodes != []:
        return nodes
    else:
        return None
```

```python
def bfs(start, end):
    # start with A
    visited = []
    Q = [start]

    # for each neighbor, add to Q
    # v_next = neighbors(start)
   #     for v in v_next:
       # Q.append(v)

    # for v in Q, append neighbors to  Q
    while Q:
        v =Q.pop(0)
        visited.append(v)
         #stop if v = end
        if v == end:
            return visited

        v_next = neighbors(v)
        if v_next != None:
            for v2 in v_next:
                if v2 not in visited :
                    Q.append(v2)

        print(v, Q)
    return 0

bfs('A','G') 
```

```python
def dfs(start, end):
    visited = []
    S = [start]

    while S:
        v = S.pop(0)
        visited.append(v)
        if v == end:
            return visited
        v_next = neighbors(v)
        if v_next != None:
            for v2 in v_next:
                if v2 not in visited:
                    S.insert(0, v2)

        print(v, S)
    return 0
dfs('A','G')
```
