```python

class edge:
  def __init__(self, start, end, weight):
    self.source= start
    self.target = end
    self.weight = weight
```

```python
edgeList = (edge('A','B',1), edge('B','D',1), edge('D','G',2), edge('B','E',3), edge('A','C',4), edge('C','F',5), edge('F','G',2))

def checkEdge(v):
  result = []
  for e in edgeList:
    if e.source == v:
      result.append(e)
  if(len(result)):
    return result
  else:
    return False


def findCost(path):
  total =0
  for i in range(0, len(path)-1):
    for e in edgeList:
      if(e.source == path[i] and e.target == path[i+1]):
        total+=(e.weight)
  return total
```

```python
def ucs(start, goal):
  paths = [[start]]
  visited = []

  while len(paths) > 0:
    paths.sort(key=findCost)
    path = paths.pop(0)
    node = path[-1]

    if node == goal:
      return path

    if node not in visited:
      visited.append(node)
      neighbours = checkEdge(node)
      if(neighbours):
        for neighbour in neighbours:
          new_path = list(path)
          new_path.append(neighbour.target)
          paths.append(new_path)

x =ucs('A','G')
print( x, findCost(x))
```

```python
h = {'A':10, 'B':9, 'C':8, 'D':7, 'E':6, 'F':5, 'G':4}

def astar(start, goal):
  paths = []
  visited = []

  paths.append([start])
  while paths:
    path = paths.pop(0)
    node = path[-1]
    if node == goal:
      return path
    elif node not in visited:
      neighbours = checkEdge(node)
      if(neighbours):
        visited.append(node)
        for neighbour in neighbours:
          new_path = list(path)
          new_path.append(neighbour.target)
          paths.append(new_path)
          paths.sort(key=lambda x: h[x[-1]])

print(astar('A','G'))

```
